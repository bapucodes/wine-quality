## 1. Architecture & Pipeline Stages
The pipeline executes 13 distinct MLOps lifecycle stages:
1. **01_load_data.py**: SHA-256 lineage tracking, batch verification, raw data archiving.
2. **02_validate_data.py**: Pandera schema validation on raw input features.
3. **03_preprocessing.py**: Stratified train/test split, StandardScaler, test ID persistence.
4. **04_validate_outputs.py**: Output boundary, shape, and NaN validation.
5. **05_train_registry.py**: Random Forest training with MLflow experiment tracking.
6. **06_model_evaluation.py**: Confusion matrix, ROC curve, and classification metrics.
7. **07_quality_gate.py**: Automated recall $\ge 0.70$ quality threshold verification.
8. **08_automate_lifecycle.py**: Model Registry champion promotion.
9. **11_prepare_deployment.py**: Deployment bundle packaging into `deployment/`.
10. **09_inference.py**: Offline batch inference execution.
11. **10_failure_injection.py**: 5 failure scenarios tested across 3 repetitions.
12. **12_data_drift.py / 12_prediction_drift.py / 12_concept_drift.py**: Multi-modal drift monitoring.
13. **12_monitor_trigger.py & 12_retrain_model.py**: Automated retraining on distribution shifts.
